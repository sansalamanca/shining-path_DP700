```
# Create spark session
from pyspark.sql import SparkSession
spark = SparkSession.builder \
    .appName("session12") \
    .getOrCreate()
```

```


# load the file from the JVM's disk (local)
df = spark.read.format("csv").option("header", "True").load("weights.csv")

df.show(5)
```

```
#  Download as CSV.

# 1. Convert the Spark DataFrame to Pandas and save it locally in Colab
nombre_csv = "weight_people.csv"
df.toPandas().to_csv(nombre_csv, index=False)

# 2. Download to your PC
from google.colab import files
files.download(nombre_csv)
```
