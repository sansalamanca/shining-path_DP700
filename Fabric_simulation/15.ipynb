# For this notebook to work, execute the cells in the established order.

```python
# Create and visualize  DF
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("session 15").getOrCreate()
# 2. Definir datos
data = [
    (101, "Ana", "Analyst", 3200.0),
    (102, "Charles", "Engineer", 4500.5),
    (103, "Helen", "Analyst", 3400.0),
    (104, "David", "Developer", 4100.0)
]
columns = ["ID", "Name", "Position", "Salary"]

# 3.Create DataFrame
df = spark.createDataFrame(data, schema=columns)

# 4. Visualize
df.show()
```

# Column selection and row filtering

```python
# 1. Selecting specific columns (.select)
from pyspark.sql.functions import col
print("--- Display only name and salary---")
df.select("Name", "Salary").show()

```

```python
# 2. Filtering rows with a condition (.filter)
print("--- Display employees with salary bigger than 3500---")
df.filter(col("Salary") > 3500).show()

```

```python
# 3. Filtro combinado (Múltiples condiciones: Analistas con salario >= 3300)
print("--- analysts with salaries higher than 3300 ---")
df.filter((col("Position") == "Analyst") & (col("Salary") >= 3300)).show()
```
