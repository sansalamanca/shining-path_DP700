# **Machine Learning Example**

# **Create a column of vectors (called features)**

```python
# 2. VectorAssembler adds a vector column where each vector contains
# the values ​​from its row, separated by commas. It will name this column 'features'.

assembler = VectorAssembler(
    inputCols=["M2", "Rooms", "Age"],
    outputCol="features"
)
df_prepared = assembler.transform(df_houses)

# 3.display DF

df_prepared.show(truncate=False)
```

**Apply LinearRegression to predict the price of a house**

```python
train_df, test_df = df_prepared.randomSplit([0.8, 0.2], seed=42)

# 5 Train the model using .fit()
# Define which algorithm we will use.

from pyspark.ml.regression import LinearRegression

# We apply the algorithm, passing it the featuresCol and labelCol columns
lr = LinearRegression(featuresCol="features", labelCol="Price_USD")
modelo_lr = lr.fit(train_df)
predicciones = modelo_lr.transform(test_df)

# Display the actual price versus the price predicted by the model
predicciones.select("ID", "features", "Price_USD", "prediction").show(truncate=False)

# PySpark randomly selected record ID 3 and calculated its price prediction.
```

# **Change the algorithm**

```python
# evaluate an existing model using RegressionEvaluator

from pyspark.ml.evaluation import RegressionEvaluator

# 1. Generate predictions for the entire DataFrame of 5 rows
total_prediction = modelo_lr.transform(df_prepared)

# 2. Configure and run the evaluator
evaluator = RegressionEvaluator(
    labelCol="Price_USD",
    predictionCol="prediction",
    metricName="r2"
)
r2 = evaluator.evaluate(total_prediction)
print(f"Coeficiente de determinación R2: {r2:.4f}")
```
