```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import sum, avg, max, min, count, col
spark = SparkSession.builder.appName("AggFunctionExample").getOrCreate()

datos = [
    ("North", "Keyboard", 50),
    ("North", "Mouse", 20),
    ("North", "Monitor", 300),
    ("South", "Laptop", 1200),
    ("South", "Keyboard", 50)
]

df = spark.createDataFrame(datos, ["region", "product", "amount"])

df.show()
```

```python
# Using agg() to calculate multiple statistics in a single pass

df_summary = df.groupBy("region").agg(
    sum(col("amount")).alias("total_amount"),
    avg(col("amount")).alias("AVG_amount"),
    max(col("amount")).alias("Highest_sale"),
    count(col("product")).alias("QTY")
)

df_summary.show()
```
