# For this notebook to work, execute the cells in the established order.

```python
import os
from pyspark.sql import SparkSession		# connects to API spark
from pyspark.sql.types import IntegerType, StringType, StructField, StructType

# 1. Iniciar la sesión de Spark
spark = SparkSession.builder.appName("session14").getOrCreate()

```

# Provide data and a schema for creating the DF

```python
# 2. Define data. Table with 3 columns
data = [
    (1, "Laptop", 1200.50),
    (2, "Mouse", 25.00),
    (3, "Keyboard", 45.99),
    (4, "Monitor", 300.00),
]

# define schema
schema = StructType(
    [
        StructField("id", IntegerType(), True),
        StructField("product", StringType(), True),
        StructField("price", DoubleType(), True)
        if "DoubleType" in globals()
        else StructField("price", StringType(), True),
    ]
)

# 'Assemble'  DF
df_original = spark.createDataFrame(data, schema)

df_original.show()   # In Fabric use display(df_original)
```

# Save the DF to the JVM's hard drive (in CSV, JSON, and Parquet formats)

```python
# Define the local path for the files. /content/data_demo
output_dir = "./data demo"

#  Save the DataFrame currently in RAM to the JVM's virtual disk (in formats CSV, Json and parquet)
df_original.write.mode("overwrite").option("header", "true").csv(f"{output_dir}/csv")
df_original.write.mode("overwrite").json(f"{output_dir}/json")
df_original.write.mode("overwrite").parquet(f"{output_dir}/parquet")
print("--- FILES SUCCESSFULLY GENERATED ---")
```

```python
# 1. Reading CSV
print("\n>>> Reading CSV:")
df_csv = (
    spark.read.option("header", "true")
    .option("inferSchema", "true")
    .csv(f"{output_dir}/csv")
)
df_csv.printSchema()
df_csv.show()

# 2. Reading JSON
print("\n>>> Reading JSON:")
df_json = spark.read.json(f"{output_dir}/json")
df_json.printSchema()
df_json.show()

# 3. Reading Parquet
print("\n>>> Reading Parquet:")
df_parquet = spark.read.parquet(f"{ output_dir}/parquet")
df_parquet.printSchema()
df_parquet.show()
```
