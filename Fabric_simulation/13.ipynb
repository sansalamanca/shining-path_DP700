```python
# 4. Registrar la vista temporal SQL necesaria para spark.sql()
df_employees.createOrReplaceTempView("employees")
```

```python
df_report = spark.sql("""
    SELECT
      Position,
      COUNT(*) AS Total_Empleados,
      ROUND(AVG(Salary), 2) AS AVG_salary
    FROM employees
    WHERE Salary > 3300
    GROUP BY Position
    ORDER BY AVG_salary DESC
""")

df_report.show()
```

```python
# 1. Convertir el DataFrame de Spark a Pandas
pandas_df = df_report.toPandas()

# 2. Guardar los archivos localmente en el entorno de Colab
pandas_df.to_csv("df_report.csv", index=False)

# Usamos openpyxl para exportar a Excel (suele venir preinstalado en Colab)
pandas_df.to_excel("df_report.xlsx", index=False, engine="openpyxl")

# 3. Descargar los archivos a tu PC
from google.colab import files
files.download("df_report.csv")
files.download("df_report.xlsx")
```
