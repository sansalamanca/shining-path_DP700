```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import udf, col
from pyspark.sql.types import StringType

spark = SparkSession.builder.appName("StandardUDF").getOrCreate()

data = [
    ("Laptop Pro", 1200),
    ("Keyboard", 80),
    ("Monitor 4K", 350)
]

df = spark.createDataFrame(data, ["product", "price"])
```

```python
# 1. Define a standard Python function

def categorize_price(price):
    if price >= 1000:
        return "high-end"
    elif price >= 200:
        return "mid range"
    else:
        return "low end"

# 2. Register the function as a UDF in PySpark specifying the return type

categorize_udf = udf(categorize_price, StringType())

# 3. Apply the UDF to the DataFrame. This will add the "category" column

outcome_df = df.withColumn("category", categorize_udf(col("price")))
outcome_df.show()
```
