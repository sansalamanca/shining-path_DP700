```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col
spark = SparkSession.builder.appName("PartitioningData").getOrCreate()
data = [
    ("2026-08-01", "Sales", 1500),
    ("2026-08-01", "IT", 2300),
    ("2026-08-02", "Sales", 1800),
    ("2026-08-02", "IT", 2100),
    ("2026-08-03", "Sales", 2000),
]
df = spark.createDataFrame(data, ["data", "department", "amount"])

# Method 1: Changing the exact number of partitions (forcing a shuffle)
df_repartitioned = df.repartition(4)
print("Partitions using the repartition method:", df_repartitioned.rdd.getNumPartitions())

# Method 2: Reduce partitions within each worker node (without a full shuffle)
df_coalesce = df_repartitioned.coalesce(2)
print("Partitions using the coalesce method:", df_coalesce.rdd.getNumPartitions())

# Method 3: Repartitioning by a specific column – Hash Partitioning. Ideal for preparing joins or groupings.
df_by_dept = df.repartition(col("department"))
print("Partitions using  department:", df_by_dept.rdd.getNumPartitions())
```
