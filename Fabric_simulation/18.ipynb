# **When the key columns have different names...**

```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Join_diff_ColumnNames").getOrCreate()

# 1. DF employees (key column called 'id_emp')
emp_data = [(1, "Ann"), (2, "Charles"), (3, "Helen"), (4, "David")]
df_emp = spark.createDataFrame(emp_data, ["id_emp", "name"])

# 2. DF deparments (key column called 'id_dept')
dept_data = [(1, "Sales"), (2, "IT"), (5, "Marketing")]
df_dept = spark.createDataFrame(dept_data, ["id_dept", "department"])
print("--- DataFrame Employees ---")
df_emp.show()

print("--- DataFrame Departments ---")
df_dept.show()    # in Fabric use:   display(df_dept)


```

```python
# EJEMPLO: INNER JOIN  Usamos la condición: df_emp.id_emp == df_dept.id_dept
df_inner = df_emp.join(df_dept, df_emp.id_emp == df_dept.id_dept, "inner")
print("--- INNER JOIN ---")
df_inner.show()   # in Fabric use:   display(df_inner)



```
