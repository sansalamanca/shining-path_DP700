# **Machine Learning Example**

# **Create a column of vectors (called features)**

```python
from pyspark.sql import SparkSession
from pyspark.ml.feature import VectorAssembler
from pyspark.ml.regression import LinearRegression

# 0. Initate Spark session
spark = SparkSession.builder \
    .appName("session 22") \
    .getOrCreate()

# 1. Create testing DF
data = [
    (1, 85, 2, 10, 150000),
    (2, 120, 3, 5, 230000),
    (3, 200, 4, 2, 410000),
    (4, 65, 1, 15, 110000),
    (5, 150, 3, 8, 290000)
]

columns = ["ID", "M2", "Rooms", "years", "Price_USD"]
df_houses = spark.createDataFrame(data, columns)

# 2. Configure and apply VectorAssembler
assembler = VectorAssembler(
    inputCols=["M2", "Rooms", "Years"],
    outputCol="features"
)
df_prepared = assembler.transform(df_houses)
df_prepared.show()      # In Fabric use:  display(df_prepared )
```

```python
# 3. Define the Linear Regression algorithm
lr = LinearRegression(featuresCol="features", labelCol="Price_USD")

# 4. Train the model using the ENTIRE DataFrame (without randomSplit)
modelo_lr = lr.fit(df_prepared)

# 5. Generate predictions for the entire DataFrame
predicciones = modelo_lr.transform(df_prepared)

# 6. Display results
predicciones.select("ID", "features", "Price_USD", "prediction").show(truncate=False)

```
