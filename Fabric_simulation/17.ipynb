# **Let´s explore the joins of data frames**

```python
#Create DF

from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("JoinExamples").getOrCreate()

# Employees DataFrame (Left)
emp_data = [(1, "Ann"), (2, "Charles"), (3, "Helen"), (4, "David")]
df_emp = spark.createDataFrame(emp_data, ["id", "name"])

# DataFrame de Departamentos (Right)
dept_data = [(1, "Sales"), (2, "IT"), (5, "Marketing")]
df_dept = spark.createDataFrame(dept_data, ["id", "department"])

# Mostrar los DataFrames
df_emp.show()
df_dept.show()
```

```python
# inner join
df_emp.join(df_dept, "id", "inner").show()
```

```python
# Left join
df_emp.join(df_dept, "id", "left").show()
```

```python
# Right join
df_emp.join(df_dept, "id", "right").show()
```

```python
#Full outer join
df_emp.join(df_dept, "id", "outer").show()
```

```python
# Left semi join
df_emp.join(df_dept, "id", "leftsemi").show()
```

```python
#Left anti join
df_emp.join(df_dept, "id", "leftanti").show()
```
