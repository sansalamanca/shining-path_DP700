```python
# Build a DF with many imperfections

import pandas as pd
import numpy as np
# Imperfect data
raw_data = {
    'client_id': [101, 102, 103, 104, 105, 106, 107, 108, 108, 109],
    'city': ['  caracas ', 'BOGOTA', ' lima', 'MADRID', None, 'CARACAS', '  bogota ', 'buenos aires', 'buenos aires', 'QUITO'],
    'sale_amount': ['150.5', '200.0', '0', '310.0', '180.0', '220.0', '0', '5000.0', '5000.0', '190.0'],
    'applied discount': [0, 10, 0, 15, np.nan, 5, 0, 20, 20, np.nan],
    'client_age': [25, 30, np.nan, 45, 38, 29, 0, 52, 52, 31]
}
df = pd.DataFrame(raw_data)
print("--- ORIGINAL DF (Dirty) ---")
print(df)
print("\n original data types:")
print(df.dtypes)
```

```python
# Create a copy to perform data cleaning
df_clean = df.copy()

# Case 1: Data type. 'sale_amount' is currently a string/object; convert it to a float.
df_clean['sale_amount'] = pd.to_numeric(df_clean['sale_amount'])

# Case 2: 'city' column (remove whitespace, convert to uppercase)
df_clean['city'] = df_clean['city'].astype(str).str.strip().str.upper()

# Restore actual NaN values ​​where 'NONE' appeared due to the conversion to string
df_clean['city'] = df_clean['city'].replace('NONE', np.nan)

# Case 3: Determine whether zeros represent legitimate values ​​or placeholders (missing data).

# In 'client_age', an age of 0 is being used as a placeholder.
# In 'applied_discount', a 0 indicates that there was legitimately no discount applied.

```

```python
# Case 4: Convert zeros to NAs (or vice versa) as appropriate
# In 'client_age', replace zeros (placeholders) with NaN so they can be handled.
# In 'applied_discount', replace NaNs with 0 (indicating the absence of a discount).
df_clean['client_age'] = df_clean['client_age'].replace(0, np.nan)
df_clean['applied discount'] = df_clean['applied discount'].fillna(0)

# Case 5: Remove rows with empty fields (if appropriate)
# If 'city' is essential for geographic analysis and is empty, remove that row. df_clean = df_clean.dropna(subset=['ciudad'])

```

```python
# Case 6: Impute (replace) NAs using statistical methods (mean or median)
# We impute missing values ​​in 'client_age' using the average/mean age.
mean_age = df_clean['client_age'].mean()
df_clean['client_age'] = df_clean['client_age'].fillna(round(mean_age))
```

```python
# Case 7: Remove exact duplicate rows (such as the row with ID 108).
df_clean = df_clean.drop_duplicates()

# Case 8: Remove outliers. Any value >1000 is considered an outlier and will be removed.
df_clean = df_clean[df_clean['sale_amount'] < 1000]
```

```python
# RESULTADO FINAL
print("--- Cleaned DATAFRAME ---")
print(df_clean)
print("\nTipos de datos finales:")
print(df_clean.dtypes)
```
