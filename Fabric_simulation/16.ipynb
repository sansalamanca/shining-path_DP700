# For this notebook to work, execute the cells in the established order.

```python
# Create and visualize  DF
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("session 15").getOrCreate()
# 2. Definir datos
data = [
    (101, "Ana", "Analyst", 3200.0),
    (102, "Charles", "Engineer", 4500.5),
    (103, "Helen", "Analyst", 3400.0),
    (104, "David", "Developer", 4100.0)
]
columns = ["ID", "Name", "Position", "Salary"]

# 3.Create DataFrame
df = spark.createDataFrame(data, schema=columns)

# 4. Visualize
df.show()
```

# **Create a new calculated column**

```python
# 1. Create a calculated column named 'Bonus' (10% of the salary). Use the (.withColumn) method.

from pyspark.sql.functions import col, avg, round
df_with_bonus = df.withColumn("Bonus", round(col("Salary") * 0.10, 2))
print("--- DATAFRAME with new column bonus ---")
df_with_bonus.show()
```

# Group by Position and calculate the average salary for each profession.

```python
# 2. Take the DF and group rows by 'Position'; calculate the sum and average of the salaries
# Round the average to two decimal places.
# Rename the resulting column to "AVG_salary"

from pyspark.sql.functions import avg, round
df_resumen = df.groupBy("Position").agg(
    round(avg("Salary"), 2).alias("AVG_salary")
)

print("--- Average salary by position ---")
df_resumen.show()



```

# **Sort results from highest to lowest salary (.orderBy)**

```python
# 3. Sort results from highest to lowest salary (.orderBy)
print("--- EMPLOYEES SORTED BY SALARY (DESCENDING) ---")
df.orderBy(col("Salary").desc()).show()
```
