# **Change names to columns**

```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("RenameColumn").getOrCreate()
data = [(1, "Ann", 30), (2, "Charles", 25)]		#  tuplas list
df = spark.createDataFrame(data, ["id", "user_name", "age"])

df.show()
```

```python
# Cambiar un solo nombre.  nombre_usuario  nombre

df_renamed = df.withColumnRenamed("user_name", "NAME")
df_renamed.show()    # In Fabric use:   display(df_renamed)
```

# **Change many column names**

```python
df_renamed_multi = df \
.withColumnRenamed("user_name", "NAME") \
.withColumnRenamed("age", "YEARS")
df_renamed_multi.show()
```
